```python
import pandas as pd
import re
from langdetect import detect
from collections import Counter
from transformers import pipeline
```

```python
file_name = "[PUBLIC] ECHA Consultation Contributions - echa_comments_with_attachments_links.csv"
df = pd.read_csv(file_name, encoding="utf-8")
# List of text columns to check (update based on your CSV's actual columns)
text_columns = [
    'General Comments',
    '1. Sectors and (sub-)uses',
    '2. Emissions in the end-of-life phase',
    '3. Emissions in the end-of-life phase',
    '4. Impacts on the recycling industry',
    '5. Proposed derogations',
    '6. Missing uses',
    '7. Potential derogations marked for reconsideration',
    '8. Other identified uses',
    '9. Degradation potential of specific PFAS sub-groups',
    '10. Analytical methods'
]
print("Columns in CSV:", df.columns.tolist())
print("\nFirst 5 rows:")
display(df.head())
```

```python
# Detect languages

# Function to detect language of first non-empty text column in a row
def get_row_language(row):
    for col in text_columns:
        text = row[col]
        if pd.notnull(text) and str(text).strip() != "":  # Skip empty or NaN
            try:
                return detect(str(text))
            except:
                continue  # Skip errors and try next column
    return 'unknown'  # If all text columns are empty

# Apply to each row
df['language'] = df.apply(get_row_language, axis=1)

# Count frequencies
language_counts = Counter(df['language'])
print("Language Frequencies:", language_counts)
```

```python
language_to_model = {
    'en': None,  # 英文无需翻译
    'sv': 'Helsinki-NLP/opus-mt-sv-en',
    'de': 'Helsinki-NLP/opus-mt-de-en',
    'it': 'Helsinki-NLP/opus-mt-it-en',
    'ca': 'Helsinki-NLP/opus-mt-ca-en',
    'fr': 'Helsinki-NLP/opus-mt-fr-en',
    'no': 'Helsinki-NLP/opus-mt-no-en',
    'nl': 'Helsinki-NLP/opus-mt-nl-en',
    'pt': 'Helsinki-NLP/opus-mt-pt-en',
    'es': 'Helsinki-NLP/opus-mt-es-en',
    'ro': 'Helsinki-NLP/opus-mt-ro-en',
    'fi': 'Helsinki-NLP/opus-mt-fi-en',
    'cs': 'Helsinki-NLP/opus-mt-cs-en',
    'tr': 'Helsinki-NLP/opus-mt-tr-en',
    'da': 'Helsinki-NLP/opus-mt-da-en',
    'sk': 'Helsinki-NLP/opus-mt-sk-en',
    'ko': 'Helsinki-NLP/opus-mt-ko-en',
    'ja': 'Helsinki-NLP/opus-mt-ja-en',
    'pl': 'Helsinki-NLP/opus-mt-pl-en',
    'so': 'facebook/m2m100_418M',  # 索马里语
    'tl': 'facebook/m2m100_418M',  # 他加禄语
    'af': 'facebook/m2m100_418M',  # 南非荷兰语
    'sq': 'facebook/m2m100_418M',  # 阿尔巴尼亚语
    'et': 'facebook/m2m100_418M',  # 爱沙尼亚语
    'sw': 'facebook/m2m100_418M',  # 斯瓦希里语
    'unknown': None  # 未知语言
}
```

```python
from transformers import pipeline
import torch

# 检查 GPU 是否可用
device = 0 if torch.cuda.is_available() else -1  # 0 = GPU, -1 = CPU

# 预加载所有翻译模型（缓存）
model_cache = {}

def get_translator(src_lang):
    """
    根据源语言代码返回对应的翻译模型
    """
    # 如果语言不在支持列表或无需翻译，返回 None
    if src_lang not in language_to_model or language_to_model[src_lang] is None:
        return None

    model_name = language_to_model[src_lang]

    # 复用已缓存的模型
    if model_name in model_cache:
        return model_cache[model_name]

    # 加载新模型
    try:
        if 'm2m100' in model_name:
            # 多语言模型需指定源语言和目标语言
            translator = pipeline(
                "translation",
                model=model_name,
                device=device,
                src_lang=src_lang,
                tgt_lang='en'
            )
        else:
            translator = pipeline(
                "translation",
                model=model_name,
                device=device
            )
        model_cache[model_name] = translator
        return translator
    except Exception as e:
        print(f"Failed to load model for {src_lang}: {e}")
        return None
```

```python
def translate_text(text, src_lang='auto'):
    """
    翻译单个文本字段
    """
    if pd.isnull(text) or str(text).strip() == "":
        return text  # 跳过空值

    # 自动检测语言
    if src_lang == 'auto':
        src_lang = get_row_language(text)

    # 获取翻译器
    translator = get_translator(src_lang)
    if translator is None:
        return text  # 无法翻译时返回原文本

    # 分段翻译（防止超长文本）
    max_length = 400
    text_segments = [str(text)[i:i+max_length] for i in range(0, len(str(text)), max_length)]
    translated_segments = []

    for segment in text_segments:
        try:
            result = translator(segment, max_length=max_length, truncation=True)
            translated_segments.append(result[0]['translation_text'])
        except Exception as e:
            print(f"Translation error for segment '{segment}': {e}")
            translated_segments.append(segment)  # 翻译失败时保留原文本

    return " ".join(translated_segments)

# 对所有文本列应用翻译
for col in text_columns:
    df[f"{col}_en"] = df.apply(lambda row: translate_text(row[col], src_lang=row['language']), axis=1)
```

```python
from transformers import pipeline
import torch

# Check if GPU is available
device = 0 if torch.cuda.is_available() else -1  # 0 = GPU, -1 = CPU

# Preload all required translation models
model_cache = {}

def get_translator(src_lang):
    if src_lang not in language_to_model or language_to_model[src_lang] is None:
        return None  # Skip translation

    model_name = language_to_model[src_lang]
    if model_name in model_cache:
        return model_cache[model_name]  # Reuse cached model

    translator = pipeline("translation", model=model_name, device=device)
    model_cache[model_name] = translator
    return translator

# Translate function with caching
def translate_text(text, src_lang='auto'):
    if pd.isnull(text) or str(text).strip() == "":
        return text

    # Detect language if not provided
    if src_lang == 'auto':
        src_lang = detect_language(text)

    translator = get_translator(src_lang)
    if translator is None:
        return text  # Return original if no model available

    # Split long texts into segments
    max_length = 400
    text_segments = [text[i:i+max_length] for i in range(0, len(text), max_length)]
    translated_segments = []

    for segment in text_segments:
        try:
            result = translator(segment, max_length=max_length, truncation=True)
            translated_segments.append(result[0]['translation_text'])
        except Exception as e:
            print(f"Translation error: {e}")
            translated_segments.append(segment)  # Fallback to original

    return " ".join(translated_segments)
```

## Attachement and confidential dummy variables

```python
def combine_text(row):
    return ' '.join(str(val).lower() for val in row[text_columns] if pd.notna(val) and val.strip())

df['combined_text'] = df.apply(combine_text, axis=1)
```

```python
def is_valid_url(url):
    if pd.isnull(url):
        return False
    pattern = r'^https?://.*'
    return bool(re.match(pattern, str(url)))

# Apply the function to the 'attachments' column
df['attachment_link_dummy'] = df['attachments'].apply(lambda x: 1 if is_valid_url(x) else 0)

# Define keywords indicating confidential references
confidential_keywords = ['confidential', 'redacted']

# Function to check for confidential references in text columns
def check_confidential(row):
    for col in text_columns:
        if pd.notnull(row[col]):
            for keyword in confidential_keywords:
                if re.search(keyword, str(row[col]), re.IGNORECASE):
                    # Check if attachments column is empty or invalid
                    if not is_valid_url(row['attachments']):
                        return 1
    return 0

# Apply the function to each row
df['confidential_info_dummy'] = df.apply(check_confidential, axis=1)

# Critical columns for sentiment analysis
critical_columns = [
    'General Comments',
    '1. Sectors and (sub-)uses',
    '5. Proposed derogations',
    '6. Missing uses',
    '7. Potential derogations marked for reconsideration'
]

def check_missing_key_info(row):
    # Step 1: Count empty critical columns
    empty_count = sum(
        pd.isna(row[col]) or str(row[col]).strip() == ''
        for col in critical_columns
    )

    # Step 2: Combine critical columns
    combined = ' '.join(
        str(val).strip()
        for val in row[critical_columns]
        if pd.notna(val) and str(val).strip()
    )

    # Step 3: Check text length
    char_count = len(combined)
    word_count = len(combined.split())

    # Step 4: Flag only if both conditions are met
    if empty_count > 2 and (char_count < 50 or word_count < 10):
        return 1  # Key info likely lost
    return 0  # No significant info loss

# Apply to DataFrame
df['missing_key_info'] = df.apply(check_missing_key_info, axis=1)
```
